## Functional API

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Dense,
    Input,
    Normalization
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.datasets import california_housing


# Load Dataset
housing = california_housing.load_data()

(X_train, y_train), (X_test, y_test) = housing

```

```python
#### Normalization layer
norm_layer=Normalization(
    name="norm_layer"
)

norm_layer.adapt(X_train)

### Input
inputs=Input(
    shape=[8],
    name="input_layer"
)

# Hidden Layers
x=norm_layer(inputs)

x=Dense(
    100, activation="relu",
    name="hidden1"
)(x)

x=Dense(
    50, activation="relu",
    name="hidden2"
)(x)

### OutPut Layer
outputs=Dense(
    1, name="output_layer"
)(x)

### Create Model
func_model=Model(
    inputs=inputs,
    outputs=outputs,
    name="functional_model"
)

# Summary
func_model.summary()

## Compile
func_model.compile(
    optimizer=Adam(0.001),
    loss="mse",
    metrics=["RootMeanSquaredError"]
)

# Train
history=func_model.fit(
    X_train, y_train, epochs=20, validation_split=0.2
)

## Evaluate
func_model.evaluate(
    X_test, y_test
)

```
