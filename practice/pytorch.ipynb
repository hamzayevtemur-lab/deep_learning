## PYTORCH vs NUMPY

```python
import numpy as np
x=np.array([1, 2, 3])
print("numpy array:", x)
```

```python
import torch
x=torch.tensor([1, 2, 3])
print("Pytorch tensor:", x)
```

```python
## ZEROS

np.zeros((3, 4)) 

torch.zeros((3, 4))
```

```python
## ONES

np.ones((3, 4))

torch.ones((3, 4))
```

```python
## FILL with constant

np.full((3, 4), 5)

torch.full((3, 4), 5)
```

```python
## Range
np.arange(1, 100, 2)

torch.arange(1, 100, 2)
```

```python
## linearly spaced values
np.linspace(1, 100, 100)

torch.linspace(1, 100, 100)
```

```python
## Random uniform
np.random.rand(2, 3)

torch.rand(2, 3)
```

```python
##Random Normal Distribution
np.random.randn(1000)

torch.randn(1000)
```

```python
## Identity Matrix
np.eye(5)

torch.eye(5)
```

```python
## Reshape
x.reshape(10, 10)

x.reshape(10, 10)
```
