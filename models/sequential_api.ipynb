## Sequential API

```python
import numpy as np
import pandas as pd

import tensorflow as tf
from tensorflow.keras.datasets import california_housing
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Dense , InputLayer, Normalization

from tensorflow.keras.optimizers import Adam
```

```python
## Load Dataset
housing=california_housing.load_data()

(X_train, y_train), (X_test, y_test)=housing

######## Create Normalization Layer
norm_layer=Normalization(
    name="norm_layer"
)
# Learn mean and std
norm_layer.adapt(X_train)



####### Build Sequential Model
seq_model=Sequential([
    InputLayer(
        shape=[8],
        name="input_layer"
    ),
    
    norm_layer,
    
    Dense(
        100,
        activation="relu",
        name="hidden1"
    ),
    
    Dense(
        100,
        activation="relu",
        name="hidden2"
    ),
    
    Dense(
        1, 
        name="output_layer"
    )
     
])

#### Model Summary
seq_model.summary()


## Compile Model
optimizer=Adam(
    learning_rate=0.001
)

seq_model.compile(
    optimizer=optimizer,
    loss="mse",
    metrics=["RootMeanSquaredError"]
)

##### Train Model
history=seq_model.fit(
    X_train, y_train, epochs=20, validation_split=0.2
)
```

```python
# Evaluate Model
seq_model.evaluate(
    X_test, y_test
)
```
