## Subclassing Api

```python
import numpy as np
import pandas as pd

import tensorflow as tf
from tensorflow.keras.datasets import california_housing
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense , InputLayer, Normalization

from tensorflow.keras.optimizers import Adam


### Load Dataset
housing=california_housing.load_data()

(X_train, y_train), (X_test, y_test)=housing
```

```python
### Custom Model

class HousingModel(Model):
    def __init__(self):
        super().__init__()
        
        # Normalization Layer
        self.norm_layer=Normalization(
            name="norm_layer"
        )
        
        # Hidden layers
        self.hidden1=Dense(
            100, 
            activation='relu',
            name="hidden1"
        )
        
        # Hidden Layer
        self.hidden2=Dense(
            100, activation="relu",
            name="hidden2"
        )
        
        # Output Layer
        self.output_layer=Dense(
            1, name="output_layer"
        )
    
        
    def call(self, inputs):
        x=self.norm_layer(inputs)
        x=self.hidden1(x)
        x=self.hidden2(x)
        
        return self.output_layer(x)
```

```python
# Create Model
sub_model=HousingModel()

# Adapt Normalization
sub_model.norm_layer.adapt(X_train)


## Build Model
sub_model.build(input_shape=(None, 8))

# Summary
sub_model.summary()


# Compile 
sub_model.compile(
    optimizer=Adam(0.001),
    loss='mse',
    metrics=["RootMeanSquaredError"]
)

## Train
history=sub_model.fit(
    X_train, y_train, epochs=20, validation_split=0.2
)

# Evaluate
sub_model.evaluate(
    X_test, y_test
)

print("\nSubclassing API:")
sub_model.evaluate(X_test, y_test)
```
